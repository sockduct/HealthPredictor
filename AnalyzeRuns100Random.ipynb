# Analyze HealthPredictor Notebook Runs

```python
# Standard Library:
from collections import defaultdict
import shelve

# 3rd Party:
from IPython.display import display as display
import pandas as pd

# Shelve Database:
DB = 'hp_results'
```

```python
# Load the shelve database
with shelve.open(DB) as db:
    # Create a dictionary to group runs by model
    model_data = defaultdict(dict)

    for full_key, metrics in db.items():
        # Split into run and model
        run, model = full_key.split('/', 1)
        model_data[model][run] = metrics

# Now convert each model's runs into a DataFrame
model_dfs = {}

for model, runs in model_data.items():
    df = pd.DataFrame.from_dict(runs, orient='index')
    df.index.name = 'RUN#'
    model_dfs[model] = df

# Now model_dfs is a dictionary: model name -> corresponding DataFrame
for model in model_dfs:
    print(f'{model} model runs:\n{'=' * 100}')
    display(model_dfs[model])
    print(f'{model} model run statistics:\n{'=' * 100}')
    display(model_dfs[model].describe())
```
